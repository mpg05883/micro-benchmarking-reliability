
# Models EDA

This notebook explores the models in the [Open LLM Leaderboard](https://huggingface.co/open-llm-leaderboard). See [here](https://huggingface.co/docs/leaderboards/open_llm_leaderboard/about) to learn more about the Open LLM Leaderboard.

## Set Up

Import packages.

```python
import ipynbname
import json
import matplotlib.pyplot as plt
import numpy as np
import os
import pandas as pd
import re
import seaborn as sns
import shutil
import warnings

from pathlib import Path

# Suppress warnings from Hugging Face
warnings.filterwarnings(
    "ignore",
    message=".*huggingface_hub.*cache-system uses symlinks.*",
)

# Suppress progress bars from Hugging Face
os.environ["HF_HUB_DISABLE_PROGRESS_BARS"] = "1"
os.environ["HF_DATASETS_DISABLE_PROGRESS_BARS"] = "1"
```

```python
from datasets import load_dataset
```

Load the Open LLM Leaderboard's [contents](https://huggingface.co/datasets/open-llm-leaderboard/contents) and filter it so it only contains models from the pre-defined filtered list.

```python
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)

cache_dir = Path.cwd() / "cache"
cache_dir.mkdir(exist_ok=True)

owner = "open-llm-leaderboard"
dataset_name = "contents"
split = "train"

contents_df = load_dataset(
    f"{owner}/{dataset_name}",
    split=split,
    cache_dir=cache_dir,
).to_pandas()

configs_dir = Path.cwd() / "configs"
with open(configs_dir / "models.json", "r") as f:
    models = json.load(f)

# Filter the leaderboard
contents_df = contents_df[contents_df["fullname"].isin(models)]

contents_df = contents_df.sort_values(
    "Average ⬆️",
    ascending=False,
).reset_index(drop=True)


print(f"Shape: {contents_df.shape}")
contents_df.head()
```

### Metadata

View the different precision types.

```python
notebook = Path(ipynbname.path()).stem
results_dir = Path.cwd() / "results" / notebook
metadata_plots_dir = results_dir / "metadata" / "plots"
metadata_plots_dir.mkdir(exist_ok=True, parents=True)

column = "Precision"
counts_df = (
    contents_df[column]
    .value_counts()
    .reset_index()
    .sort_values(
        by="count",
        ascending=False,
    )
)

fig, ax = plt.subplots(figsize=(10, 4))
sns.barplot(
    x=counts_df[column],
    y=counts_df["count"],
    ax=ax,
    hue=counts_df[column],
    palette="tab10",
    legend=False,
)

# Annotate bars with their heights
for container in ax.containers:
    ax.bar_label(container, fontweight="bold")

title = f"{column.title()} Types"
plt.title(title)
plt.xlabel(f"{column.title()} Type")
plt.ylabel("Number of Models")

file_name = f"{title.lower().replace(' ', '_')}__barplot.png"
path = metadata_plots_dir / file_name
plt.savefig(path)

plt.show()
```

### Model Scores

View the distribution of average scores across all models.

```python
scores_plots_dir = results_dir / "scores" / "plots"
scores_plots_dir.mkdir(exist_ok=True, parents=True)

column = "Average ⬆️"
data = contents_df[column].dropna()

idx_max = contents_df[column].idxmax()
idx_min = contents_df[column].idxmin()

print(
    f"Model with the highest average: {contents_df.loc[idx_max, 'fullname']} "
    f"({contents_df.loc[idx_max, column]:.2f})"
)
print(
    f"Model with the lowest average: {contents_df.loc[idx_min, 'fullname']} "
    f"({contents_df.loc[idx_min, column]:.2f})"
)

max_val = data.max()
fig, ax = plt.subplots(figsize=(10, 5))
bins = np.linspace(0, max_val, 11)  # Use 11 edges to get 10 bins
sns.histplot(data=data, bins=bins, ax=ax)
ax.set_xlim(0, max_val)

# Display median
median_val = data.median()
ax.axvline(
    median_val,
    color="red",
    linestyle="--",
    linewidth=2,
    label=f"Median: {median_val:.2f}",
)

# Annotate each bar's height
for patch in ax.patches:
    if patch.get_height() > 0:
        ax.text(
            patch.get_x() + patch.get_width() / 2,
            patch.get_height(),
            f"{int(patch.get_height())}",
            ha="center",
            va="bottom",
            fontweight="bold",
        )

ax.legend()

# Remove emoji
column = "Average"

title = f"Distribution of {column.title()} Across All Benchmarks"
ax.set_title(title)
ax.set_xlabel(column.title())
ax.set_ylabel("Number of Models")

file_name = f"{column.lower().replace(' ', '_')}__histogram.png"
path = scores_plots_dir / file_name
plt.savefig(path)

plt.show()
```

View the distribution of model performance on each benchmark.

```python
benchmarks = [
    "IFEval",
    "BBH",
    "MATH Lvl 5",
    "GPQA",
    "MUSR",
    "MMLU-PRO",
]

benchmarks_to_metrics = {
    "IFEval": "Accuracy",
    "BBH": "Accuracy",
    "MATH Lvl 5": "Exact Match",
    "GPQA": "Accuracy",
    "MUSR": "Accuracy",
    "MMLU-PRO": "Accuracy",
}

colors = sns.color_palette("tab10")

annotation_fontsize = 12
tick_fontsize = annotation_fontsize + 2
legend_fontsize = tick_fontsize
label_fontsize = tick_fontsize + 2
title_fontsize = tick_fontsize + 2
suptitle_fontsize = title_fontsize + 2

raw_annotation_color = "red"
normalized_annotation_color = "teal"
linestyle = "--"
linewidth = 3
alpha = 0.3

for benchmark in benchmarks:
    raw_column = f"{benchmark} Raw"
    normalized_column = benchmark

    fig, axes = plt.subplots(1, 2, figsize=(20, 5), sharey=True)

    tuples = [
        ("Raw", raw_column, (0, 1)),
        ("Normalized", normalized_column, (0, 100)),
    ]

    for idx, (score_type, column, xlim) in enumerate(tuples):
        ax = axes[idx]

        if column not in contents_df.columns:
            ax.set_visible(False)
            continue

        data = contents_df[column].dropna()

        is_raw = score_type == "Raw"
        max_val = 1 if is_raw else 100
        bins = np.linspace(0, max_val, 11)
        sns.histplot(
            data=data,
            bins=bins,
            ax=ax,
            color=colors[idx],
        )
        ax.set_xlim(*xlim)

        # Annotate each bar's height
        for patch in ax.patches:
            if patch.get_height() > 0:
                ax.text(
                    patch.get_x() + patch.get_width() / 2,
                    patch.get_height(),
                    f"{int(patch.get_height())}",
                    ha="center",
                    va="bottom",
                    fontweight="bold",
                    fontsize=annotation_fontsize,
                )

        # Display median
        median_val = data.median()

        annotation_color = (
            raw_annotation_color if is_raw else normalized_annotation_color
        )

        ax.axvline(
            median_val,
            color=annotation_color,
            linestyle=linestyle,
            linewidth=linewidth,
            label=f"Median: {median_val:.2f}",
        )

        q1 = data.quantile(0.25)
        q3 = data.quantile(0.75)
        iqr = q3 - q1

        # Display IQR as shaded region
        ax.axvspan(
            q1,
            q3,
            alpha=alpha,
            color=annotation_color,
            label=f"IQR: {iqr:.2f}",
        )

        metric = benchmarks_to_metrics[benchmark]

        ax.legend(fontsize=legend_fontsize)
        ax.tick_params(labelsize=tick_fontsize)
        ax.set_title(f"{score_type} {metric} Scores", fontsize=title_fontsize)
        ax.set_xlabel(f"Score (0–{max_val})", fontsize=label_fontsize)
        ax.set_ylabel("Number of Models", fontsize=label_fontsize)

    suptitle = f"Distribution of {benchmark} Scores\nAggregated Across All Subsets"
    plt.suptitle(
        suptitle,
        fontsize=suptitle_fontsize,
        y=1.01,
    )
    plt.tight_layout()

    plot_name = benchmark.lower().replace(" ", "_").replace("-", "_")
    file_name = f"{plot_name}__histogram.png"
    path = scores_plots_dir / file_name
    plt.savefig(path)

    plt.show()
```
