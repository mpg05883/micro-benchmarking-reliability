# Leaderboard EDA

This notebook explores the [Open LLM Leaderboard](https://huggingface.co/open-llm-leaderboard). See [here](https://huggingface.co/docs/leaderboards/open_llm_leaderboard/about) to learn more about the Open LLM Leaderboard.

## Set Up

Import packages.

```python
import os
import pandas as pd
import shutil
import time
import warnings

from pathlib import Path

# Suppress warnings from Hugging Face
warnings.filterwarnings(
    "ignore",
    message=".*huggingface_hub.*cache-system uses symlinks.*",
)

# Suppress progress bars from Hugging Face
os.environ["HF_HUB_DISABLE_PROGRESS_BARS"] = "1"
os.environ["HF_DATASETS_DISABLE_PROGRESS_BARS"] = "1"
```

```python
from datasets import load_dataset
from huggingface_hub import list_datasets
```

Get the metadata for each dataset in the Open LLM Leaderboard from Hugging Face.

```python
author = "open-llm-leaderboard"
max_attempts = 5

for attempt in range(max_attempts):
    try:
        datasets = list(list_datasets(author=author))
        print(f"Attempt {attempt + 1}: Success!")
        break
    except Exception as e:
        print(f"Attempt {attempt + 1} failed. {type(e).__name__}")
        if attempt < max_attempts - 1:
            time.sleep(2**attempt)  # Exponential backoff
        else:
            raise
```

View each dataset's metadata.

```python
num_preview = 10

print(f"Found {len(datasets)} datasets:")
for i, dataset in enumerate(datasets[:num_preview]):
    print(f"{i + 1}. {dataset}")
```

## Open LLM Leaderboard 2

The [Open LLM Leaderboard 2](https://huggingface.co/collections/open-llm-leaderboard/open-llm-leaderboard-2) is one of the Open LLM Leaderboard's collections, and contains the following datasets:
- [`open-llm-leaderboard/contents`](https://huggingface.co/datasets/open-llm-leaderboard/contents) -  Aggregated results for the Open LLM Leaderboard 
- [`open-llm-leaderboard/results`](https://huggingface.co/datasets/open-llm-leaderboard/results) -  Detailed per-benchmark evaluation data for all the evaluations we run
- [`open-llm-leaderboard/requests`](https://huggingface.co/datasets/open-llm-leaderboard/requests) -  Status request of all the models submitted
- [`open-llm-leaderboard/official-providers`](https://huggingface.co/datasets/open-llm-leaderboard/official-providers) -  The shortlist of model creators we run in priority

Load the Open LLM Leaderboard's [contents](https://huggingface.co/datasets/open-llm-leaderboard/contents) to view the leaderboard.

```python
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)

cache_dir = Path.cwd() / "cache"
cache_dir.mkdir(exist_ok=True)

dataset_name = "contents"
split = "train"

contents_df = load_dataset(
    f"{author}/{dataset_name}",
    split=split,
    cache_dir=cache_dir,
).to_pandas()

contents_df = contents_df.sort_values(
    "Average ⬆️",
    ascending=False,
).reset_index(drop=True)


print(f"Shape: {contents_df.shape}")
contents_df.head()
```

## Clean Up

Clear the cache directory.

```python
if cache_dir.exists():
    shutil.rmtree(cache_dir)
    print(f"Cleared cache at {cache_dir}")
else:
    print("No cache to clear")
```
