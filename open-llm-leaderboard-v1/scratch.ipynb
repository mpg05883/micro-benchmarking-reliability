```python
from datasets import load_dataset
from datasets.config import HF_DATASETS_CACHE
import os

# Increase timeout and retry settings for large datasets
os.environ["HF_HUB_DOWNLOAD_TIMEOUT"] = "300"  # 5 minutes timeout
os.environ["HF_HUB_DISABLE_SYMLINKS_WARNING"] = "1"

# V1 aggregated results - use streaming first to avoid connection issues
# Then convert to pandas
v1_results = load_dataset(
    "open-llm-leaderboard-old/results",
    split="train",
    cache_dir=cache_dir,
    num_proc=1,  # Single process to avoid connection flooding
)
v1_df = v1_results.to_pandas()

v1_df.head()
```

```python
# Alternative: Use streaming mode (more resilient to connection issues)
# This downloads data on-demand rather than all at once

from datasets import load_dataset
import pandas as pd

# Stream the dataset - this won't download everything upfront
v1_stream = load_dataset(
    "open-llm-leaderboard-old/results",
    split="train",
    streaming=True,
)

# Convert streamed dataset to a list, then to pandas
# This processes in batches and is more resilient to connection drops
rows = []
for i, row in enumerate(v1_stream):
    rows.append(row)
    if i % 1000 == 0:
        print(f"Loaded {i} rows...")

v1_df = pd.DataFrame(rows)
print(f"Total rows: {len(v1_df)}")
v1_df.head()
```
